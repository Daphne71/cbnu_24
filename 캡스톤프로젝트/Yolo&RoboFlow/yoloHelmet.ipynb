```python
from ultralytics import YOLO

# 경로 설정 (필요시 수정)
dataset_yaml = "yoloHelmet.yaml"

model = YOLO("yolo11n.yaml")
model = YOLO("yolo11n.pt")
model = YOLO("yolo11n.yaml").load("yolo11n.pt")

# Train the model
model.train(data="yoloHelmet.yaml", epochs=50, imgsz=640, name="yolov11_helmet")

val_results = model.val()

print("/n Validation Metrics")
print(f"mAP50:      {val_results.box.map50:.4f}")
print(f"mAP50-95:   {val_results.box.map:.4f}")
```

```python
import cv2
import os

os.makedirs("helmet_bbox_result", exist_ok=True)
model = YOLO("runs/detect/yolov11_helmet/weights/best.pt")  # load a custom model
results = model.predict(source="helmetDataset/test/images", conf=0.1)

for result in results:
    image = result.orig_img.copy()
    boxes = result.boxes

    for box in boxes:
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        conf = float(box.conf[0])
        class_id = int(box.cls[0])
        label = f"{result.names[class_id]} {conf:.2f}"

        # 바운딩 박스와 텍스트 추가
        cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(image, label, (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    # 결과 저장
    filename = os.path.basename(result.path)
    saved = cv2.imwrite(f"helmet_bbox_result/{filename}", image)
```
